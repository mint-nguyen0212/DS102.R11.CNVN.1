## $f(x,y)=x^2+2y^2-4x-8y+10$

```python
import numpy as np

def gradient(v):
  x, y = v
  return np.array([
      2*x - 4,
      4*y - 8
  ])
```

```python
def hessian(v):
  return np.array([
      [2,0],
      [0,4]
  ])
```

```python
def newton_raphson(gradient, hessian, x0, tol=1e-6, max_lap=100):
  x = np.array(x0)
  for i in range(max_lap):
    g = gradient(x)
    if np.linalg.norm(g) < tol:
      return x, i
    h = hessian(x)
    nhay = np.linalg.solve(h, -g)
    x = x + nhay
  return x, max_lap
```

```python
if __name__ == "__main__":
  point, iterations = newton_raphson(gradient, hessian, [6, -2])
  print(point, iterations)
```
